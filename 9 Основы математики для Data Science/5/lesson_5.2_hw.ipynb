```python
from sympy import *
import numpy as np
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt
from sympy.solvers.inequalities import solve_univariate_inequality
```

```python
x = Symbol('x')
f1 = 3*x + 5
f2 = 0.25*x**2 + 0.75*x + 1.25
f3 = 0.5*x**3 - 0.75*x + 1.25
```

```python
Y_true = [12.258, 17.24, 30.151]
Y_pred = [f1.subs(x, 2), f1.subs(x, 4), f1.subs(x, 8)]
MSE = mean_squared_error(Y_true, Y_pred)
print(MSE)

x_range = np.linspace(0, 10, 100)
fu1 = lambda x: 3*x + 5
plt.scatter([2,4,8],[12.258,17.24,30.151], c='r')
plt.axhline(0, color='black')
plt.axvline(0, color='black')
plt.plot(x_range, fu1(x_range))
```

```python
Y_true = [3.688, 10.791, 20.705]
Y_pred = [f2.subs(x, 2), f2.subs(x, 4), f2.subs(x, 8)]
MSE = mean_squared_error(Y_true, Y_pred)
print(MSE)

x_range = np.linspace(0, 10, 100)
fu2 = lambda x: 0.25*x**2 + 0.75*x + 1.25
plt.scatter([2,4,8], Y_true, c='r')
plt.axhline(0, color='black')
plt.axvline(0, color='black')
plt.plot(x_range, fu2(x_range))
```

```python
Y_true = [4.872, 29.707, 246.971, 485.727, 840.658]
Y_pred = [f3.subs(x, 2), f3.subs(x, 4), f3.subs(x, 8), f3.subs(x, 10), f3.subs(x, 12)]
MSE = mean_squared_error(Y_true, Y_pred)
MSE2 = np.square(np.subtract(Y_true,Y_pred)).mean() 
print(MSE)
print(MSE2)

x_range = np.linspace(0, 14, 100)
fu3 = lambda x: 0.5*x**3 - 0.75*x + 1.25
plt.scatter([2,4,8,10,12], Y_true, c='r')
plt.axhline(0, color='black')
plt.axvline(0, color='black')
plt.plot(x_range, fu3(x_range))
```
