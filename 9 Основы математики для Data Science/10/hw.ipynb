```python
import numpy as np
from sympy import *
```

```python
x, y, z = symbols('x, y, z')
a, b = symbols('a, b')
x1, x2, x3 = symbols('x1, x2, x3')
```

```python
eq1 = -1*x + 7*y + 4
eq2 = 8*x + 8*y + 48
linsolve([eq1, eq2], [x,y])
```

```python
eq1 = 4*x - 7*y + 4
eq2 = 3*x - 4*y + 3
linsolve([eq1, eq2], [x,y])
```

```python
eq1 = 8*a - 4*b - 64
eq2 = -3*a + 3*b + 21
linsolve([eq1, eq2], [a,b])
```

```python
eq1 = 5*x1 + 7*x2 - 5*x3 + 47
eq2 = -2*x2 + 2*x3 - 10
eq3 = -4*x1 - 8*x2 - 7 * x3 - 63
linsolve([eq1, eq2, eq3], [x1,x2,x3])
```

```python
MSE = 1/6 * ((5*x + 7*y -5*z + 47)**2 + \
             (-2*y + 2*z - 10)**2 + \
             (-4*x -8*y -7*z - 63)**2 + \
             (x + y + 2*z + 1)**2 + \
             (2*x - y + 2*z + 4)**2 + \
             (4*x + y + 4*z + 2)**2)

MSE.subs({x:0, y:0, z:0})
```

```python
MSEx = diff(MSE, x)
MSEy = diff(MSE, y)
MSEz = diff(MSE, z)
step = 0.01
```

```python
tochka = np.array([0, 0, 0])
while MSE.subs({x: tochka[0], y: tochka[1], z: tochka[2]}) > 55:
    obj = {x: tochka[0], y: tochka[1], z: tochka[2]}
    odMSE = -1*step*np.array([MSEx.subs(obj), MSEy.subs(obj), MSEz.subs(obj)])
    tochka = tochka + odMSE
    print(tochka)
    print(MSE.subs({x: tochka[0], y: tochka[1], z: tochka[2]}))
```
