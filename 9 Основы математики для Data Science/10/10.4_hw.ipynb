```python
from sympy import *
```

```python
x, y, z = symbols('x, y, z')

eq1 = 2*x + 5*y - 11
eq2 = 7*x - 3*y + 23
```

```python
linsolve([eq1, eq2], [x,y])
```

```python
eq1 = -3*x + y + 2
eq2 = 3*x + 5*y - 8
linsolve([eq1, eq2], [x,y])
```

```python
eq1 = 2*x + 3*y - 12
eq2 = 3*x - y - 7
linsolve([eq1, eq2], [x,y])
```

```python
eq1 = x + y + 2*z + 1
eq2 = 2*x - y + 2*z + 4
eq3 = 4*x + y + 4*z + 2
linsolve([eq1, eq2, eq3], [x,y,z])
```
