```python
from sympy import *
from sympy.plotting import plot
from sympy.plotting import plot3d
init_printing(use_unicode=False, wrap_line=False, no_global=True)
x, y = symbols('x, y')
```

```python
f = x*(y-5)**2
f
```

```python
fx = diff(f, x)
print(fx)
plot(fx)
```

```python
fy = diff(f, y)
print(fy)
plot3d(fy)
```

```python
f2 = (y - E/20)**2 * sin(pi * x)
f2
```

```python
f2x = diff(f2, x)
print(f2x)
plot3d(f2x)
```

```python
f2y = diff(f2, y)
print(f2y)
plot3d(f2y)
```

```python
f3 = x*sin(x)*sin(y-2)
f3
```

```python
f3x = diff(f3, x)
print(f3x)
plot3d(f3x)
```

```python
f3y = diff(f3, y)
print(f3y)
plot3d(f3y)
```

```python
a2, a1, a0 = symbols('a2, a1, a0')
```

```python
MSE = 1/3 * (((a2 * 1945 + a1 * 560 + a0) - 4.3)**2 + \
             ((a2 * 1495 + a1 * 340 + a0) - 4.9)**2 + \
             ((a2 * 1570 + a1 * 343 + a0) - 5.2)**2)
MSE
```

```python
MSEa2 = diff(MSE, a2)
MSEa1 = diff(MSE, a1)
MSEa0 = diff(MSE, a0)
```

```python
p = nonlinsolve([MSEa2, MSEa1, MSEa0],[a2, a1, a0])
print(p)
```
