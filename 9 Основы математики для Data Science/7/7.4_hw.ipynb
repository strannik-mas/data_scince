```python
from sympy import *
from sympy.plotting import plot3d
init_printing(use_unicode=False, wrap_line=False, no_global=True)
```

```python
x, y = symbols('x, y')
```

```python
f1 = 2*x + 5*y + 1
f1
```

```python
solve(f1)
```

```python
f2 = -5*y/2 - 1/2
plot(f2)
```

```python
plot3d(f1)
```

```python
f1 = x**2 + y**2 - 1
f1
```

```python
solve(f1)
```

```python
f2 = -1*sqrt(1 - y**2)
```

```python
f3 = sqrt(1 - y**2)
plot(f2, (y, -1, 1))
```

```python
plot(f3, (y, -1, 1))
```

```python
plot3d(f1)
```

```python
solve(f1)
```
