```python
X = Matrix([[ 5.0,  7.0, -5.0],
            [ 0.0, -2.0,  2.0],
            [-4.0, -8.0, -7.0],
            [ 1.0,  1.0,  2.0],
            [ 2.0, -1.0,  2.0],
            [ 4.0,  1.0,  4.0]])
X
```

```python
b = Matrix([-47.0, 10.0, 63.0, -1.0, -4.0, -2.0])
b
```

```python
X_pseudo_inverse = (X.T * X)**(-1) * X.T
X_pseudo_inverse
```

```python
best_point = X_pseudo_inverse * b
best_point
```

```python
MSE = 1 / 6 * ((5 * x + 7 * y - 5 * z + 47)**2 + \
               (- 2 * y + 2 * z - 10)**2 + \
               (-4 * x - 8 * y - 7 * z - 63)**2 + \
               (x + y + 2 * z + 1)**2 + \
               (2 * x - y + 2 * z + 4)**2 + \
               (4 * x + y + 4 * z + 2)**2)
MSE.subs({x: best_point[0], y: best_point[1], z: best_point[2]})
```
