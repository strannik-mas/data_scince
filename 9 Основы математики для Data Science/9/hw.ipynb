```python
def transponirovanieMatrix(matrix):
    listTmp = []
    razm = matrix.shape
    for i in range(razm[1]):
        intListTmp = []
        for j in range(razm[0]):
            intListTmp.append(matrix[j][i])
        listTmp.append(intListTmp)
    return np.array(listTmp)
```

```python
M = np.array([[2,1,7,4],[5,6,7,3],[9,8,2,12],[11,14,15,15]])
Mt = transponirovanieMatrix(M)
Mt
```

```python
M2 = np.array([[3,7,8,3,6],[2,5,9,4,13]])
M2t = transponirovanieMatrix(M2)
M2t
```

```python
a1, a2 = symbols('a1, a2')

MSE = 1 / 4 * ((1 * a1 + 2 * a2 - 5)**2 + \
               (5 * a1 + 3 * a2 - 6)**2 + \
               (2 * a1 + 4 * a2 - 10)**2 + \
               (3 * a1 + 7 * a2 - 8)**2)
MSE.subs({a1: 0.57, a2: 0.91})
```

```python
MSEa1 = diff(MSE, a1)
MSEa1
```

```python
MSEa2 = diff(MSE, a2)
MSEa2
```

```python
step = 0.01
```

```python
deltaA1 = MSEa1.subs({a1: 0.57, a2: 0.91})
deltaA1
```

```python
deltaA2 = MSEa2.subs({a1: 0.57, a2: 0.91})
deltaA2
```

```python
tochka = np.array([0.57, 0.91])
odMSE = -1* step* np.array([deltaA1, deltaA2])
odMSE
```

```python
tochka2 = tochka + odMSE
tochka2
```
