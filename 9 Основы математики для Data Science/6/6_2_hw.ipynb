```python
from sympy import *
import numpy as np
import matplotlib.pyplot as plt
from sympy.solvers.inequalities import solve_univariate_inequality
```

```python
x = Symbol('x')
f1 = x**2 + 3*x - 4
f1_proizv = 2*x + 3
```

```python
solve(f1_proizv)
```

```python
solve_univariate_inequality(f1_proizv > 0, x)
```

```python
solve_univariate_inequality(f1_proizv < 0, x)
```

```python
solve(f1)
```

```python
x_range = np.linspace(-5, 5, num=100)
fu = lambda x: x**2 + 3*x - 4
fu2 = lambda x: 2*x + 3
plt.axhline(0, color='black')
plt.axvline(0, color='black')
plt.plot(x_range, fu(x_range))
plt.plot(x_range, fu2(x_range))
```
