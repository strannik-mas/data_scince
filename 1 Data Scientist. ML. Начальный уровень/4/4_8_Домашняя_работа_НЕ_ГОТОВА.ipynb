```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.datasets import fetch_openml
from scipy.spatial import distance
from sklearn.linear_model import SGDRegressor
from sklearn.metrics import r2_score
%matplotlib inline

boston_dataset = fetch_openml(name='boston', version=1)
# X = boston_dataset.data
X = boston_dataset.data['CRIM'].values.reshape(-1, 1)
# X['CHAS'] = X['CHAS'].astype(float)
# X['RAD'] = X['RAD'].astype(float)
y = boston_dataset.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=10)
print(X_train.shape)
m = X.shape[1]
# X_train
# -- ВАШ КОД ТУТ --
sgd_regressor = SGDRegressor(learning_rate='constant', eta0=0.015, fit_intercept=True, random_state=42)
w_current, epsilon = np.random.random(2), 0.0001
weight_evolution, r2_evolution = [], [] # изменения весов и ошибка на валидации
for step in list(range(800)):
    sgd_regressor = sgd_regressor.partial_fit(X_train, y_train)
    euclidean = distance.euclidean(w_current, sgd_regressor.coef_)
    print(euclidean)
    weight_evolution.append(euclidean)
    if step == 5: break
    # print(weight_evolution[-1])
    if weight_evolution[-1]<epsilon:
        print("Итарации остановлены на шаге %d" % step); break
    r2_evolution.append(r2_score(y_train, sgd_regressor.predict(X_train)))
    w_current = sgd_regressor.coef_.copy()
# plt.plot(range(step), r2_evolution)
# print(len(r2_evolution))
# l2 = range(step)
# print(len(l2))
```
